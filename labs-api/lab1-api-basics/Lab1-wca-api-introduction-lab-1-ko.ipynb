### Lab 1.1. Swagger 명세로부터 REST API 생성

Watson Code Assistant (WCA) API와 상호작용하여, 주어진 프롬프트와 Swagger (OpenAPI 명세) 파일을 기반으로 파이썬 코드를 생성하는 과정을 보여줍니다. Flask 프레임워크를 사용하여 최소한의 기본 웹 API 만들고 있고 @app.route(URL 엔드포인트 정의)와 jsonify(JSON 응답을 생성하는 함수)만을 사용하도록 프롬프트를 작성하였습니다.
아래 셀을 실행하면 결과를 확인할 수 있으며, 다른 prompt나 Swagger 명세로 자유롭게 실험해 볼 수 있습니다.

```python
# sample
text = """Create a minimal Flask API using only @app.route and jsonify, based on the given Swagger:""" # 프롬프트

payload = build_basic_prompt_paylod(text)
response = call_wca_url(payload=payload, apikey=iam_apikey, file_dict=["assets/petstore-minimal.json"]) # Swagger 파일
result = (response['response']['message']['content'])
print(result)
```
